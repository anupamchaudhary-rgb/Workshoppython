```python
# ==========================================================
# SIMPLIFIED DIGITAL TWIN OF A CNC MACHINE
# ==========================================================

import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
data = pd.read_csv("cnc_digital_twin_sample_data.csv")

# Display first five rows
print("FIRST FIVE ROWS OF THE DATASET")
display(data.head())

# ----------------------------------------------------------
# GRAPH 1: SPINDLE SPEED
# ----------------------------------------------------------

plt.figure(figsize=(8, 4))

plt.plot(
    data["Machining Time (min)"],
    data["Spindle Speed (RPM)"]
)

plt.title("Spindle Speed vs Machining Time")
plt.xlabel("Machining Time (min)")
plt.ylabel("Spindle Speed (RPM)")
plt.grid()
plt.show()

# ----------------------------------------------------------
# GRAPH 2: TOOL TEMPERATURE
# ----------------------------------------------------------

plt.figure(figsize=(8, 4))

plt.plot(
    data["Machining Time (min)"],
    data["Tool Temperature (°C)"]
)

plt.title("Tool Temperature vs Machining Time")
plt.xlabel("Machining Time (min)")
plt.ylabel("Tool Temperature (°C)")
plt.grid()
plt.show()

# ----------------------------------------------------------
# GRAPH 3: TOOL WEAR
# ----------------------------------------------------------

plt.figure(figsize=(8, 4))

plt.plot(
    data["Machining Time (min)"],
    data["Tool Wear (mm)"]
)

plt.title("Tool Wear vs Machining Time")
plt.xlabel("Machining Time (min)")
plt.ylabel("Tool Wear (mm)")
plt.grid()
plt.show()

# ----------------------------------------------------------
# CALCULATE STATISTICS
# ----------------------------------------------------------

maximum_temperature = data["Tool Temperature (°C)"].max()

average_spindle_speed = data["Spindle Speed (RPM)"].mean()

maximum_tool_wear = data["Tool Wear (mm)"].max()

print("\nIMPORTANT RESULTS")

print(
    "Maximum Tool Temperature:",
    round(maximum_temperature, 2),
    "°C"
)

print(
    "Average Spindle Speed:",
    round(average_spindle_speed, 2),
    "RPM"
)

print(
    "Maximum Tool Wear:",
    round(maximum_tool_wear, 3),
    "mm"
)

# ----------------------------------------------------------
# DETECT ABNORMAL CONDITIONS
# ----------------------------------------------------------

high_temperature = data[
    data["Tool Temperature (°C)"] > 90
]

high_vibration = data[
    data["Vibration (mm/s)"] > 3
]

high_tool_wear = data[
    data["Tool Wear (mm)"] > 0.35
]

print("\nHIGH TEMPERATURE RECORDS")
display(high_temperature)

print("\nHIGH VIBRATION RECORDS")
display(high_vibration)

print("\nHIGH TOOL WEAR RECORDS")
display(high_tool_wear)

# ----------------------------------------------------------
# CALCULATE REMAINING USEFUL LIFE
# ----------------------------------------------------------

maximum_allowable_wear = 0.50
wear_rate = 0.011
current_wear = maximum_tool_wear

remaining_tool_life = (
    maximum_allowable_wear - current_wear
) / wear_rate

print(
    "\nRemaining Useful Life:",
    round(remaining_tool_life, 2),
    "minutes"
)

# ----------------------------------------------------------
# DETERMINE MACHINE STATUS
# ----------------------------------------------------------

if current_wear >= 0.50:

    machine_status = "Replace Tool Immediately"

elif current_wear >= 0.45:

    machine_status = "Critical Condition"

elif current_wear >= 0.35:

    machine_status = "Maintenance Required"

else:

    machine_status = "Machine is Healthy"

print("Machine Status:", machine_status)
```
