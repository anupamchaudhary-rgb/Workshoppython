```python
units = input("Enter the number of units consumed: ")
units = int(units)
customer_type = input("Enter customer type (residential, commercial, industrial): ").lower()

price = 0

if customer_type == "residential":
    price = units * 2
elif customer_type == "commercial":
    price = units * 5
elif customer_type == "industrial":
    price = units * 10
else:
    print("Invalid customer type entered.")
    price = 0 # Or handle as an error condition

if price > 0:
    print(f"For {units} units and a {customer_type} customer, the total energy price is: ${price:.2f}")
```

```python
num=input("enter the number")
num=int(num)
i=1 #Initialize Variable
while(i<=10): #While(condition):
  print(f"{num} * {i} = {num*i}")
  if(i==5):
    break
  i=i+1 #Increment Variable Value
```

```python
List_of_Books=["Python","Java","C","C++"]
book=input("enter book you want in library")
book_no=0 #Start from index 0
while(book_no<len(List_of_Books)): #Loop condition corrected to avoid out of bound access
  if(book==List_of_Books[book_no]):
    print("yes,book available")
    break
  book_no=book_no+1
```
