```python
import cv2

# Read the image
image = cv2.imread("washers.jpg")

# Convert the image into grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Convert the grayscale image into black and white
_, binary = cv2.threshold(gray, 180, 255, cv2.THRESH_BINARY_INV)

# Find all contours
contours, hierarchy = cv2.findContours(
    binary,
    cv2.RETR_CCOMP,
    cv2.CHAIN_APPROX_SIMPLE
)

washer_count = 0
good_washers = 0
defective_washers = 0

# Check every contour
for i, contour in enumerate(contours):

    # Outer contours have no parent
    if hierarchy[0][i][3] == -1:

        area = cv2.contourArea(contour)

        # Ignore very small objects
        if area > 1000:

            washer_count = washer_count + 1

            x, y, w, h = cv2.boundingRect(contour)

            # Check whether the washer has an inner contour
            child = hierarchy[0][i][2]

            if child != -1:
                good_washers = good_washers + 1
                colour = (0, 255, 0)
                text = "GOOD"
            else:
                defective_washers = defective_washers + 1
                colour = (0, 0, 255)
                text = "NO HOLE"

            # Draw a rectangle around the washer
            cv2.rectangle(
                image,
                (x, y),
                (x + w, y + h),
                colour,
                3
            )

            # Write result above the washer
            cv2.putText(
                image,
                text,
                (x, y - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                colour,
                2
            )

print("Total washers =", washer_count)
print("Good washers =", good_washers)
print("Defective washers =", defective_washers)

# Show the result
cv2.imshow("Washer Inspection", image)

# Save the inspected image
cv2.imwrite("inspection_result.jpg", image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
